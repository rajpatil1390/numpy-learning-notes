# **7 Boolean & Logical Function**

```python
# np.all() checks whether all values in the array are True.
# If even one value is False, result will be False.
logical_test = np.array([True, False, True])
all_true = np.all(logical_test)  # Check if all are True
print("All elements True:", all_true)
```

```python
logical_test = np.array([True, True, True])
all_true = np.all(logical_test)  # Check if all are True
print("All elements True:", all_true)
```

```python
# np.any() checks whether at least one value in the array is True.
# If one or more values are True, result will be True.
logical_test1 = np.array([False,False,False])
any_true = np.any(logical_test1)
print("Any elements True:", any_true)
```

```python
any_true = np.any(logical_test)
print("Any elements True:", any_true)
```

# **8 Set Operation**

```python
set_a = np.array([1,2,3,4])
set_b = np.array([3,4,5,6])
intersection = np.intersect1d(set_a,set_b)
print("Intersection of a and b:", intersection)
```

```python
union = np.union1d(set_a, set_b)
print("Union of a and b:", union)
```
