```python
def generate_otp(length=4): # Define a function named generate_otp with default length 4
   
    """Generate a numeric OTP of a specified length.""" # Docstring: explains what this function does
    digits = '0123456789' # Store allowed digits for OTP; only 0,1,2,3,4,5 will be used
    otp = ''.join(random.choice(digits) for _ in range(length)) # Choose random digits and join them into one OTP string
    return otp # Return the generated OTP value


otp_length = 4  # Set OTP length to 4 digits
otp = generate_otp(otp_length) # Call generate_otp function and store returned OTP in otp variable
print(f"Your OTP is: {otp}") # Print the final OTP using f-string
```
