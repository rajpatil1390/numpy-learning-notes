# **Reshape The Aarray** 

```python
a = np.array([0,1,2,3,4,5])
a 
```

```python
a.reshape(3,2)
```

```python
a.reshape(6,1)
```

```python
a.reshape(1,6)
```

```python
a.reshape(2,3)
```

```python
a.reshape(2,4)
```

```python
arr = np.array([0,1,2,3,4,5])  # 6 elements

arr.reshape(3,2)   # -> 3 rows, 2 cols  (3x2=6 OK!)
arr.reshape(6,1)   # -> 6 rows, 1 col
arr.reshape(1,6)   # -> 1 row, 6 cols
arr.reshape(2,3)   # -> 2 rows, 3 cols
print(arr)
```

```python
a.reshape(2,3,order = 'C')  # Reshape array into 2 rows and 3 columns; default order is 'C' means row-wise filling 
```

```python
a.reshape(2,3,order = 'F') # Reshape array into 2 rows and 3 columns using Fortran order (column-wise filling)
```

```python
a.reshape(2,3,order = 'A') # Reshape array into 2 rows and 3 columns; 'A' uses Fortran order if array is Fortran-contiguous, otherwise C order
```
